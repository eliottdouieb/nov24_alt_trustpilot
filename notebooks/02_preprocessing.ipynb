# Import et chargement des données

```python
# === Standard ===
import re
import pickle
from pathlib import Path

# === Data ===
import numpy as np
import pandas as pd

# === NLP ===
import spacy
from tqdm import tqdm

# Fallback language detection.
# fastText / cld3 donnent de meilleurs résultats, mais posent des problèmes d'installation sur certaines machines.
from langdetect import detect, DetectorFactory, LangDetectException

# === ML ===
from sklearn.preprocessing import normalize
from sklearn.feature_extraction.text import TfidfVectorizer

# === Embeddings ===
from sentence_transformers import SentenceTransformer
from gensim.models import Word2Vec, KeyedVectors

# === Chargement de modèles Spacy ===
from spacy.util import is_package
import sys
import subprocess
```

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data" / "raw"

df = pd.read_csv(DATA_DIR / "reviews_trust.csv")
df.head()
```

# Preprocessing Général

```python
# Suppression des doublons et des commentaires vides ou trop courts
df = df.dropna(subset=["Commentaire"])
df = df.drop_duplicates(subset=['Commentaire', 'client'], keep='first')

def count_words(text):
    text = re.sub(r'[^\w\s]', '', text)
    return len(text.split())

df = df[df['Commentaire'].apply(count_words) > 3]

df.shape
```

```python
# Nettoyage du texte avant la vectorisation
def clean_text(text):
    text = text.lower().strip()

    # Suppression des éléments inutiles à l'aide de regex
    text = re.sub(r"(http|www)\S+", "", text) # URLs
    text = re.sub(r"\S+@\S+", "", text) # emails
    text = re.sub(r"(@|#)\w+", "", text) # mentions @ et #
    text = re.sub(r"\s+", " ", text).strip() # espaces multiples, sauts de lignes, tabulations

    return text

df["clean_comment"] = df["Commentaire"].apply(clean_text)
```

```python
# Suppression des commentaires pas en français
DetectorFactory.seed = 42

def is_french(text: str) -> bool:
    if not text or len(text.strip()) < 10:
        return False

    try:
        return detect(text) == "fr"
    except LangDetectException:
        return False

df = df[df["clean_comment"].apply(is_french)].reset_index(drop=True)
df.shape
```

```python
SPACY_MODEL = "fr_core_news_md"

def load_spacy_model(model_name: str):
    """
    Télécharge et charge un modèle spaCy si nécessaire.
    """
    if not is_package(model_name):
        print(f"Téléchargement du modèle spaCy : {model_name}")
        subprocess.check_call([
            sys.executable, "-m", "spacy", "download", model_name
        ])
    return spacy.load(model_name, disable=["ner", "parser"])

nlp = load_spacy_model(SPACY_MODEL)
```

```python
# Traitement Spacy -> lemmatisation, suppression des stopwords, des tokens de moins de 3 caractères et des tokens qui ne contiennent pas de lettres

def preprocess_spacy(doc) -> list[str]:
    return [
        token.lemma_
        for token in doc
        if token.is_alpha and not token.is_stop and len(token) > 2
    ]

docs = list(
    nlp.pipe(
        df["clean_comment"],
        batch_size=64
    )
)

df["clean_tokens"] = [preprocess_spacy(doc) for doc in docs]
```

```python
STOPWORDS_CUSTOM = {
    'montre', 'montres', 'boucle', 'boucles', 'oreilles', 'oreille', 'paire', 'paires', 'lunettes', 'bracelet', 'bracelets', 'collier', 'iphone', 'téléphone', 'pendentif', 'robot', 'robots', 'aspirateur', 'baskets', 'basket', 'chaussure', 'chaussures', 'sandales', 'plantes', 'plante', 'arbres', 'arbre', 'bulbes', 'willemse', 'sommiers', 'jardin', 'bague', 'lampe', 'lampes', 'abat-jour', 'abat jour', 'lampadaire', 'parfum', 'shampoing', 'shampooing', 'shampooings', 'cheveux', 'masque', 'masques', 'crème', 'élastiques', 'manteau', 'bougie', 'cadre', 'écouteurs', 'vélo', 'robe', 'vêtements', 'bijoux', 'sac', 'portable', 'clio', 'luminaire', 'oreillette', 'induction', 'écouteur', 'couette', 'samsung', 'téléphones', 'smartcase', 'abat', 'apple', 'watch', 'shirt', 'tee', 'chemise', 'shirts', 'hortensias', 'orchidée', 'sacs', 'plant', 'chaises', 'lacoste', 'polo', 'pantalon', 'jean', 'jeans', 'sneakers', 'lunette', 'écran', 'tablette', 'tablettes', 'table', 'tables', 'chemisier', 'pulls', 'pull', 'trotinettes', 'trotinette', 'chaussons', 'chausson', 'brosse', 'brosses', 'crèmes', 'gel', 'gels', 'parfums', 'robe', 'robes', 'sacoches', 'sacoche', 'vestes', 'veste'
}

df["comment_sans_contexte"] = df["clean_tokens"].apply(
    lambda tokens: " ".join(
        t for t in tokens if t not in STOPWORDS_CUSTOM
    )
)
```

# Segmentation en phrases

```python
def is_valid_sentence(s):
    return bool(re.search(r"[A-Za-zÀ-ÖØ-öø-ÿ]", s))

def split_into_sentences(doc):
    return [sent.text.strip() 
            for sent in doc.sents 
            if is_valid_sentence(sent.text.strip())]
```

```python
docs = list(tqdm(nlp.pipe(df["clean_comment"], batch_size=64, n_process=8), total=len(df))) # à adapter selon les coeurs du CPU et la RAM de la machine

df_sentences = df.copy()
df_sentences["comment_id"] = df.index 
df_sentences["sentences_list"] = [split_into_sentences(doc) for doc in docs]

df_sentences = df_sentences.explode("sentences_list")

df_sentences = df_sentences.rename(columns={"sentences_list": "sentence"})

df_sentences["sentence"] = df_sentences["sentence"].astype(str).str.strip()
df_sentences = df_sentences[df_sentences["sentence"].apply(count_words) > 3]

df_sentences = df_sentences.reset_index(drop=True)

# Résultat
print(f"Nombre de lignes dans df : {len(df)}")
print(f"Nombre de lignes dans df_sentences : {len(df_sentences)}")
```

```python
with open("../data/processed/dataset_final_phrases.pkl", "wb") as f:
  pickle.dump(df_sentences, f)
```

# Embedding TF-IDF

```python
# Application du preprocessing
df["clean_comment_tfidf"] = df["clean_tokens"].apply(lambda tokens: " ".join(tokens))
```

```python
# Vectorisation TF-IDF
# Suppression des tokens qui apparaissent dans plus de 95% des documents et dans moins de 5 documents
vectorizer = TfidfVectorizer(max_df=0.95, min_df=5)
tfidf_embeddings = vectorizer.fit_transform(df["clean_comment_tfidf"])
tfidf_embeddings = normalize(tfidf_embeddings, norm='l2')

# Si on veut afficher la shape des embeddings
# tfidf_embeddings.shape
```

```python
# A execiuter seulement si on veut afficher les mots les plus importants selon le TF-IDF moyen
feature_names = vectorizer.get_feature_names_out()
tfidf_means = tfidf_embeddings.mean(axis=0).A1  # moyenne par colonne
top_indices = tfidf_means.argsort()[::-1][:20]

print("Mots les plus importants (TF-IDF moyen) :")
for i in top_indices:
    print(feature_names[i], tfidf_means[i])
```

# Word embeddings

Executez cette cellule avant de passer aux cellules suivantes de words embeddings.

```python
def sentence_embedding_from_word_vectors(
    tokens: list[str],
    model,
    vector_size: int,
    get_vector_fn
) -> np.ndarray:
    """
    Calcule un embedding de phrase par moyenne des vecteurs de mots.
    """
    vectors = [
        get_vector_fn(model, token)
        for token in tokens
        if get_vector_fn(model, token) is not None
    ]

    if not vectors:
        return np.zeros(vector_size)

    return np.mean(vectors, axis=0)


sentences = df["clean_tokens"].tolist()
```

## Modèle Word2Vec entraîné avec nos données

```python
W2V_VECTOR_SIZE = 300

model_w2v_custom = Word2Vec(
    sentences,
    vector_size=W2V_VECTOR_SIZE,
    sg=1,
    workers=4
)

def w2v_get_vector(model, token):
    return model.wv[token] if token in model.wv else None

w2v_custom_embeddings = np.array([
    sentence_embedding_from_word_vectors(
        tokens, model_w2v_custom, W2V_VECTOR_SIZE, w2v_get_vector
    )
    for tokens in sentences
])

w2v_custom_embeddings = normalize(w2v_custom_embeddings, norm="l2")

# Si on veut afficher la shape des embeddings
# w2v_custom_embeddings.shape
```

```python
with open(Path("../data/embeddings/emb_w2v_custom.pkl"), "wb") as f:
    pickle.dump(w2v_custom_embeddings, f)
```
