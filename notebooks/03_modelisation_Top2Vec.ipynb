Entraînement Top2Vec (Kinjal)

Imports libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

from top2vec import Top2Vec
```

```python
# Chargement du dataset généré par le notebook de preprocessing
df = pd.read_pickle(Path("../data/processed/dataset_final.pkl"))
```

```python
# Pour le modèle Top2Vec, on utilise la colonne clean_comment
documents = df["clean_comment"].astype(str).tolist()
documents
```

Entraînement du modèle Top2Vec

```python
model = Top2Vec(
    documents=documents,
    embedding_model="distiluse-base-multilingual-cased", # embeddings transformers
    speed="learn",
    workers=4
)
```

Nombre de topics détectés

```python
print("Number of topics:", model.get_num_topics())
```

Exploration des topics

```python
topic_words, word_scores, topic_nums = model.get_topics(3)

for i, words in enumerate(topic_words[:5]):
    print(f"Topic {i + 1}: {', '.join(words)}") # Chaque topic est représenté par ses mots les plus significatifs.
```

```python
# Taille des topics
topic_sizes, topic_nums = model.get_topic_sizes()

topic_df = pd.DataFrame({
    "Topic": topic_nums,
    "Nombre d'avis": topic_sizes
}).sort_values(by="Nombre d'avis", ascending=False)

topic_df.head(10)
```

Visualisation 

```python
# Visualisation - diagramme en barres
plt.figure(figsize=(10, 6))
plt.bar(topic_df["Topic"], topic_df["Nombre d'avis"])
plt.xlabel("Numéro du Topic")
plt.ylabel("Nombre d'avis")
plt.title("Top2Vec – Nombre d'avis par topic")
plt.show()
```

```python
# Visualisation 
plt.figure(figsize=(8, 8))
plt.pie(
    topic_df["Nombre d'avis"],
    labels=topic_df["Topic"],
    autopct="%1.1f%%"
)
plt.title("Top2Vec – Proportion des avis par topic")
plt.show()
```
